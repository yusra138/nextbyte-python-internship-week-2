```python

import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime, timedelta
```

```python
# 1. Grocery Budget Planner

def grocery_budget():
    budget = float(budget_input.value)

    if budget <= 0:
        raise ValueError("Budget must be greater than zero.")

    total = 0

    for i in range(3):
        price = float(grocery_prices[i].value)
        quantity = int(grocery_quantities[i].value)

        if price < 0 or quantity <= 0:
            raise ValueError("Enter a valid price and quantity.")

        total += price * quantity

    print("Total Grocery Cost:", total, "PKR")
    print("Remaining Budget:", budget - total, "PKR")

    if total > budget:
        print("Warning: You exceeded your budget!")


# 2. Electricity Bill Estimator

def electricity_bill():
    rate = float(rate_input.value)

    if rate <= 0:
        raise ValueError("Electricity rate must be positive.")

    total_units = 0

    for i in range(3):
        wattage = float(wattage_inputs[i].value)
        hours = float(hours_inputs[i].value)

        if wattage <= 0 or hours <= 0 or hours > 24:
            raise ValueError("Enter valid wattage and daily hours.")

        units = (wattage * hours * 30) / 1000
        total_units += units

    print("Monthly Units:", round(total_units, 2), "kWh")
    print("Estimated Bill:", round(total_units * rate, 2), "PKR")


# 3. Pantry Expiry Manager

def pantry_expiry():
    today = datetime.today().date()

    for i in range(3):
        name = food_names[i].value.strip()

        if not name:
            raise ValueError("Food name cannot be empty.")

        expiry = datetime.strptime(
            expiry_inputs[i].value, "%Y-%m-%d"
        ).date()

        days = (expiry - today).days

        if days < 0:
            status = "EXPIRED"
        elif days == 0:
            status = "EXPIRES TODAY"
        elif days <= 3:
            status = "USE SOON"
        else:
            status = "SAFE"

        print(name, ":", status)


# 4. Water Tank Calculator

def water_calculator():
    capacity = float(capacity_input.value)
    current = float(current_input.value)
    usage = float(usage_input.value)

    if capacity <= 0 or usage <= 0:
        raise ValueError("Capacity and usage must be positive.")

    if current < 0 or current > capacity:
        raise ValueError("Current water must be between 0 and capacity.")

    days = current / usage

    print("Remaining Water:", current, "Litres")
    print("Estimated Supply:", round(days, 1), "days")

    if current <= capacity * 0.20:
        print("WARNING: Water level is low!")


# 5. Household Chore Planner

def chore_planner():
    completed = 0
    pending = 0

    for i in range(3):
        task = chore_names[i].value.strip()

        if not task:
            raise ValueError("Task name cannot be empty.")

        status = chore_statuses[i].value

        print(task, "-", status)

        if status == "Completed":
            completed += 1
        else:
            pending += 1

    print("Completed Tasks:", completed)
    print("Pending Tasks:", pending)
```

```python
# Main heading

heading = widgets.HTML("""
<h1 style="color:#2563eb">HomeSaver Hub</h1>
<p>Smart Household Problem Solver</p>
""")

problem = widgets.Dropdown(
    options=[
        "Grocery Budget Planner",
        "Electricity Bill Estimator",
        "Pantry Expiry Manager",
        "Water Tank Calculator",
        "Household Chore Planner"
    ],
    description="Select Problem:"
)

# Grocery inputs
budget_input = widgets.FloatText(description="Budget (PKR):", value=5000)

grocery_prices = [
    widgets.FloatText(description=f"Item {i+1} Price:")
    for i in range(3)
]

grocery_quantities = [
    widgets.IntText(description=f"Item {i+1} Quantity:", value=1)
    for i in range(3)
]

# Electricity inputs
rate_input = widgets.FloatText(description="Rate per Unit:", value=50)

wattage_inputs = [
    widgets.FloatText(description=f"Appliance {i+1} Watts:")
    for i in range(3)
]

hours_inputs = [
    widgets.FloatText(description=f"Daily Hours {i+1}:")
    for i in range(3)
]

# Pantry inputs
food_names = [
    widgets.Text(description=f"Food {i+1}:")
    for i in range(3)
]

expiry_inputs = [
    widgets.Text(description=f"Expiry {i+1}:", placeholder="YYYY-MM-DD")
    for i in range(3)
]

# Water inputs
capacity_input = widgets.FloatText(description="Tank Capacity:")
current_input = widgets.FloatText(description="Current Water:")
usage_input = widgets.FloatText(description="Daily Usage:")

# Chore inputs
chore_names = [
    widgets.Text(description=f"Task {i+1}:")
    for i in range(3)
]

chore_statuses = [
    widgets.Dropdown(
        options=["Pending", "Completed"],
        description=f"Status {i+1}:"
    )
    for i in range(3)
]

# Output area and button
input_area = widgets.VBox()

run_button = widgets.Button(
    description="Solve Problem",
    button_style="success"
)

output_area = widgets.Output()

display(heading, problem, input_area, run_button, output_area)
```

```python
def update_gui(change):
    selected = problem.value

    if selected == "Grocery Budget Planner":
        fields = [budget_input]
        for i in range(3):
            fields.extend([grocery_prices[i], grocery_quantities[i]])

    elif selected == "Electricity Bill Estimator":
        fields = [rate_input]
        for i in range(3):
            fields.extend([wattage_inputs[i], hours_inputs[i]])

    elif selected == "Pantry Expiry Manager":
        fields = []
        for i in range(3):
            fields.extend([food_names[i], expiry_inputs[i]])

    elif selected == "Water Tank Calculator":
        fields = [capacity_input, current_input, usage_input]

    else:
        fields = []
        for i in range(3):
            fields.extend([chore_names[i], chore_statuses[i]])

    input_area.children = tuple(fields)


def solve_problem(button):
    with output_area:
        clear_output()

        try:
            if problem.value == "Grocery Budget Planner":
                grocery_budget()

            elif problem.value == "Electricity Bill Estimator":
                electricity_bill()

            elif problem.value == "Pantry Expiry Manager":
                pantry_expiry()

            elif problem.value == "Water Tank Calculator":
                water_calculator()

            else:
                chore_planner()

        except ValueError as error:
            print("Input Error:", error)


problem.observe(update_gui, names="value")
run_button.on_click(solve_problem)

update_gui(None)
```
